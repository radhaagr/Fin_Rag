```python
from langchain_community.document_loaders import PyPDFLoader
```

```python
pdf_path = "./EU AI Act.pdf"

loader = PyPDFLoader(pdf_path)
documents = loader.load()

print("Total number of pages loaded:", len(documents))
```

```python
print("\n--- FIRST PAGE CONTENT ---\n")
print(documents[0].page_content[:1000])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python

text_splitter = RecursiveCharacterTextSplitter( chunk_size=1000,      # max characters per chunk
                                                chunk_overlap=200     # overlap helps preserve context
                                              )
chunks = text_splitter.split_documents(documents)
print("Total number of chunks:", len(chunks))
```
