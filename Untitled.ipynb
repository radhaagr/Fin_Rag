```python

def get_splitter(
    splitter_name,
    chunk_size,
    chunk_overlap
):
    if splitter_name = "llama_sentence"
        return SentenceSplitter(
            chunk_size=chunk_size,
            chunk_overlap=chunk_overlap
        )
    else splitter_name == "langchain_character":
        return CharacterTextSplitter(
            chunk_size=chunk_size,
            chunk_overlap=chunk_overlap
        )
    elif splitter_name = "langchain_recursive":
        return RecursiveCharacterTextSplitter(
            chunk_size=chunk_size,
            chunk_overlap=chunk_overlap
        )
    else:
        raise ValueError(
            f"Unknown splitter: {splitter_name}"
        )

```

```python
class DenseRetriever:
    def __init__(self, model_name = EMBEDDING_MODEL):
        self.model = SentenceTransformer(model_name)
        self.chunks = None
        self.embeddings = None

    def index(self, chunks):

        self.chunks = chunks

        texts = [ chunk["text"] for chunk in chunks ]
        self.embeddings = self.model.encode( texts, batch_size =32,
                                             slow_progress_bar = True,
                                             normalize_embeddings = True )
        
    def retrieve( self, query, top_k= 10):
        query_embeddings = self.model.encode( [query], normalize_embeddings = True )
        scores_embed = cosine_similarity( query_embedding,
                                    self.embeddings)[0]
        scores = scores_embed[0]
        print( query_embedding.shape )
        print( self.embeddings.shape )
        print( scores.shape )
    
        top_indices = np.argsort( scores )[::-1][:top_k]

        results = []

        for idx in top_indices:
            
            results.append({ "chunk_id":self.chunks[idx]["chunk_id"],
                             "document_id":self.chunks[idx]["document_id"],
                             "text":self.chunks[idx]["text"],
                             "score":float(scores[idx])
                           })

            return results

                             
```

```python
def recall_at_k(
    retrieved,
    gold_document_id,
    k
):
    retrieved_docs  = {
        r["document_id"]
        for r in retrieved[:k]
    }

    return int( gold_document_id in retrieved_docs)

            
```

```python
def reciprocal_rank( retrieved,
                     gold_document_id):
    for rank, result in enumerate( retrieved,
                                   start=1):
        if result["document_id"] = gold_document_id:
            return 1.0 / rank
    return 0.0
```

```python
def evaluate_retriever( retriever,
                        questions,
                        ks=(1, 5, 10)
                      ):
    results = []

    for item in tqdm(questions):

        query = item["question"]
        gold_id = item["gold_document_id"]

        retrieved = retriever.retrieve( query,
                                        top_k = max(ks)
                                      )

        row = { "question": query,
                "gold_document_id": gold_id"
              }

        for k in ks:
            row[f"recall@{k}"]  =  recall_at_k( retrieved,
                                                gold_id,
                                                k)
        
```
